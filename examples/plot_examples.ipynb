```python
import numpy as np
import matplotlib.pyplot as plt
from brypy import util, plot
```

```python
arrays = [np.random.rand(100, 100) for _ in range(8)]

plot.set_aas_style()

fig, axes = plt.subplots(2, 4, figsize=(plot.TWO_COLUMN_WIDTH, 3.5))
for ax, array in zip(axes.flatten(), arrays):
    ax.imshow(array)
plt.savefig('two.png')
plt.show()
```

```python
plt.figure(figsize=(plot.ONE_COLUMN_WIDTH, 3))
plt.imshow(arrays[0])
plt.colorbar()
plt.savefig('single.png')
plt.show()
```

```python
x = np.linspace(0, 2 * np.pi, 100)

plt.figure(figsize=(plot.ONE_COLUMN_WIDTH, 2.5))
for i in range(10):
    plt.plot(x, np.sin(x + i * np.pi / 10), label=f'Wave {i+1}')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.legend()
plt.savefig('sine_waves.png')
plt.show()
```
